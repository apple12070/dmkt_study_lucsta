```python
######### 판다스 예제 ##########
import pandas as pd

# 경로설정
PATH = "COVID-19-master/csse_covid_19_data/csse_covid_19_daily_reports/"
doc = pd.read_csv(f"{PATH}04-01-2020.csv",encoding="utf-8-sig")
```

```python
# head 함수로 최초 5개의 레코드 가져오기! -> EDA 탐색적 데이터 분석과정 (head가 어떤형태를 띄고 있는지 보기위한 과정)
doc.head()
# 데이터의 의미
# FIPS (Federal Information Processing Standards) 정보 처리 관련 표준의 집합, 미국 내 각 주, 군, 시를 숫자로 구분하는 코드 체계
```

```python
# Country_Region의 값만 찾아보고싶을때 # 데이터프레임내 틀정 컬럼만 찾아옴
countries = doc["Country_Region"]
#countries
# object -> 문자열 or 데이터형태 2개가 섞여있는 것

countries.head() #5개만 추려서 보기
```

```python
# 결측치가 아닌 정상값을 가지고 있는 데이터의 크기, countries의 크기, 규모를 보기 (count해서 몇개가 있는)
print(countries.size)
```

```python
#결측치(Null or 비어있는 값) 보기 count -> 결측치를 제외하고 제대로 있는 값의 수만 가져옴! (error의 소지가 없음)
print(countries.size,countries.count())
```

```python
#unique -> 중복값을 제거하고 대표하는 하나의 값(유일값)을 추출해서 리스트형태로 보는 방법 = 어떠한 나라들이 있는지 한번에 보고 싶을때!
print(countries.unique())
```

```python
# .value_counts() -> 각 유일값의 중복값(대표값)을 알아보는방법 = 각국이 몇번씩 언급되었는지
countries.value_counts()
# 미국 중심으로 코로나가 연구되었음을 확인할 수 있음!
```

```python
# 코로나 통계의 핵심 값만 추출해서 covid_stat에 값을 넣기 
# 확진자, 사망자, 회복 보기
covid_stat = doc[["Confirmed","Deaths","Recovered"]] # 각각의 시리즈를 데이터프레임의 형태로 가져오는 방법 = 중첩된 대괄호! 
covid_stat.head()
# doc 에서 각 3개의 컬럼만 추출해서 값을 보여줌
```

```python
## 데이터 합치기
# SQL 의 GROUP BY 
doc = pd.read_csv(f"{PATH}04-01-2020.csv",encoding="utf-8-sig")
doc.head()

# 상대적으로 어떤 차이가 있는지 유의미한 데이터를 도출하기 위해선 여러가지 데이터를 합칠 필요가 있음
# 국가별 -> 위의 size, unique 등은 보기위한것! 전처리하지 않음 (데이터자체가 변하지는 않음)

doc = doc.groupby("Country_Region").sum() # groupby 함수 = SQL, 합하려면 집계함수가 필요하기에 sum() 추가!! 
doc.head() # 그룹화시켜주면 Country_Region이 알아서 인덱스화 됨. # sum() : 숫자로 된 데이터들을 모두 합해줌!
```

```python
# 각각의 필드에 들어가는 컬럼들이 일원화되어있는 값으로 되는 문제점을 해결하는 방법
############################## NEW CHAPTER #############################

import pandas 
PATH = "COVID-19-master/csse_covid_19_data/csse_covid_19_daily_reports/"
```

```python
# 컬럼타입 변경
doc = pd.read_csv(f"{PATH}01-22-2020.csv",encoding="utf-8-sig")
doc = doc[["Country/Region","Confirmed"]] # 두 컬럼값을 가지고 있는 데이터 프레임을 생성, doc에 값을 재할당

#doc.info() # .info() -> 위에서 만든 데이터프레임의 "명세서"를 출력! 명세서(dtypes,columns,memory) # 날짜가 float의 형태!
# 4월 22일 <-> 1월 22일 데이터의 날짜의 형태가 다름! float <-> int
# 테이블을 합치려면 하나의 형태로 변환해야함!
doc= doc.dropna(subset = ["Confirmed"]) # Confirmed를 부분집합으로 가져와서 NaN이 있으면 날림 -> NaN인 10개의 국가가 날아감! 
doc = doc.astype({"Confirmed":"int64"}) #데이터의 결측치에 타입에 별칭을 준다 = Confirmed의 데이터 타입을 int로 변환한다.
doc.info()
```

```python
doc = pd.read_csv(f"{PATH}01-22-2020.csv",encoding="utf-8-sig")
doc = doc[["Country/Region","Confirmed"]]
#doc = doc.astype({"Confirmed":"int64"})# 결측치의 값을 버리지않고 타입을 변환하려할때 오류가남!
# -> 순서가 중요하다! 먼저 결측치를 날리고 타입변환해줘야 제대로 작동!
doc= doc.dropna(subset = ["Confirmed"])
doc = doc.astype({"Confirmed":"int64"})
doc.info()
```

```python
################################## NEW CHAPTER ##########################
# 중복되어진 값을 그룹화하는게 아닌 제거하고 싶다면? (groupby는 중복되는 값을 압축해놓은것!)
# 데이터 프레임에서 중복 행 확인 / 제거
import pandas as pd
PATH = "COVID-19-master/csse_covid_19_data/UID_ISO_FIPS_LookUp_Table.csv" # 데이터 새로 찾아오기
```

```python
doc = pd.read_csv(PATH,encoding="utf-8-sig")
# US 대표하는거 하나만 찾아오기! (그룹핑X)
doc = doc[["iso2","Country_Region"]] #필요한 컬럼만 찾아와서 데이터프레임 가져오기 # iso2 -> 국가,지역별 줄임 ID
doc
```

```python
# 중복여부 판별
doc.duplicated() # duplicated() -> 해당 인덱스의 값이 중복된 값이 존재하는지 판별해주는 함수!
```

```python
doc[doc.duplicated()] #true인 값 하나만 끄집어내서 데이터프레임 형태로 가져오기
```

```python
# 현재 여러번 중복해서 사용된 값들을 1개만 남기고 나머지는 제거
# 어떤 값을 살리고, 어떤 기준으로 나머지의 값을 제거할것인지 정의가 필요!
doc = doc.drop_duplicates(subset = "Country_Region", keep = "last") # 컬럼을 부분집합을 만들고(기준으로) 중복되어지는 값은 버려라
# keep first or last 무엇을 남길것인지 정하는 함수
doc # 3353개에서 180개로 하나로 줄어듬!
```
