```python
import pandas as pd

df1 = pd.DataFrame({
    "id": [1, 2, 3],
    "customer_id": [1, 2, 3],
    "customer_name": ["Robert", "Peter", "Dave"]
}, columns=["id", "customer_id", "customer_name"])

df1
```

```python
df2 = pd.DataFrame({
    "id": [1, 2, 4],
    "order_id": [100, 200, 300],
    "order_date": ["2025-01-21", "2025-02-03", "2025-09-01"]
}, columns=["id", "order_id", "order_date"])

df2
```

```python
# 서로 다른 데이터 프레임을 하나로 합치는 방법

# 1) concat() : 서로 다른 데이터 프레임을 단순 연결 (*교차되는 특정 인덱스를 고려 x) -> 좌.우 & 상.하 (디폴트)
# 2) merge() : 특정 인덱스를 기준
```

```python
pd.concat([df1, df2])
```

```python
doc = pd.concat([df1, df2], axis=1)
doc
```

```python
pd.merge(df1, df2) # merge() => inner JOIN 기능 유사
```

```python
pd.merge(df1, df2, on="id")
```

```python
pd.merge(df1, df2, on="id", how="inner")
```

```python
pd.merge(df1, df2, on="id", how="outer")
```

```python
pd.merge(df1, df2, on="id", how="left")
```

```python
pd.merge(df1, df2, on="id", how="right")
```

```python
df1 = df1.set_index("id")
df1
```

```python
df2 = df2.set_index("id")
df2
```

```python
pd.merge(df1, df2, left_index=True, right_index=True)
```

```python
pd.merge(df1, df2, how="outer", left_index=True, right_index=True)
```

```python
pd.merge(df1, df2, how="left", left_index=True, right_index=True)
```

```python
pd.merge(df1, df2, how="right", left_index=True, right_index=True)
```

```python
df = pd.DataFrame({
    "영어": [60, 70],
    "수학": [100, 50],
}, columns=["영어", "수학"], index=["Dave", "David"])

df
```

```python
df = pd.DataFrame({
    "성별" : ["남", "남", "남"],
    "이름" : ["David", "Dave", "Romeo"],
    "수학" : [100, 50, 80],
    "국어" : [80, 70, 50]
})
df
```

```python
selected_columns = ["이름","수학","국어"]

df = df[selected_columns]
# df.groupby("이름").sum()
df.groupby("이름").mean()  # mean() 평균값을 구하는 함수
```
