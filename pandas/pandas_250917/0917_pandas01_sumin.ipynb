```python
# NaN 데이터를 버릴 때???
# NaN = Not a Number (=숫자가 아님) 컴퓨터는 우리가 쓰는 모든 문자를 숫자로 인식허자먼! (*ASCII 코드 참고)
# => 측정할 수 없는 값 (insnull로 찾아냄!)

doc = pd.read_csv(f"{PATH}04-01-2020.csv", encoding="utf-8-sig")
doc.isnull()  # .isnull() : 항상은 아니지만 거의 대부분 불리언 값을 가져옴!!! True = null임
```

#### **GROUPBY** 나옴... 문법이 중고급으로 넘어가고 있다는 뜻~
- 위에서 한 size, unique, counts 이런 애들은 명세서처럼 데이터를 보여 주기만 하는 거고 전처리가 된 건 아님 !!!
- 이제 할 것.
- 그룹화가 되면 *집계*할 수 있다!

```python
doc = pd.read_csv(f"{PATH}04-01-2020.csv", encoding="utf-8-sig")
doc.head()
```

```python
doc = doc.groupby("Country_Region").sum()
doc.head()
# 그룹화가 되면 *집계*할 수 있다!
# 그룹화가 되면 자동으로 인덱스가 됨
```

```python
print(doc.columns) # 컬럼들을 보고 싶을 때 (첫 행)
print(doc.index) # 인덱스를 보고 싶을 때 (첫 열) # 아까 groupby를 줬기 때문에 그게 자동으로 인덱스가 됨
```

```python
# 미국을 찾고 싶을 때! index 안에 US가 있을 테니까!
doc[doc.index == "US"]
# doc의 index 값 중에 "US"에 부합하는 값을 ""doc"" 안에서 찾아와! (그래서 doc 안에 대괄호표기법을 쓰는 것이얌!)
```

```python
# 📌 컬럼 타입 변경

import pandas as pd
PATH = "COVID-19-master/csse_covid_19_data/csse_covid_19_daily_reports/"
```

```python
doc = pd.read_csv(f"{PATH}01-22-2020.csv", encoding="utf-8-sig")
doc = doc[["Country/Region", "Confirmed"]] # 1. 특정 컬럼만 가져오기
doc.info()
# Confirmed는 실수자료형 형태 근데!!! 04-01 데이터의 Confirmed는 정수 형태임...!!!! 일괄 형태로 바꿔주지 않으면 큰 오차/에러 생길 것임 ㅠ!!! 그래서 바꿔주는 거
```

```python
# 결측치는 버리고 시작함... dropna! *근데 Country/Region에는 결측치가 없으니 Confirmed만 부분집합으로 지정해줌!
doc = doc.dropna(subset="Confirmed") # >>>>>>> *중요 이 과정이 없으면 에러남!!!!!! 특정 컬럼 값을 변경하기 전에 결측치가 존재하면 안 됨!!!!!
doc = doc.astype({"Confirmed": "int64"}) # .astype() 특정 컬럼의 타입을 변경!
doc.info()
# Confirmed의 결측치 10개를 버렸기 때문에 나라 데이터 10개도 같이 버려져서 76에서 66로 변경됐고, 타입은 float64에서 int64로 변경됐음...!!! 이게 전처리 과정 !!!
# 왜 Confirmed의 결측치를 0으로 변경하지 않고 drop을 하느냐! -> 정보의 신뢰도가 떨어짐
```

```python
# 📌 데이터프레임에서 중복 행 확인/제거

import pandas as pd

PATH = "COVID-19-master/csse_covid_19_data/UID_ISO_FIPS_LookUp_Table.csv"
```

```python
doc = pd.read_csv(PATH, encoding="utf-8-sig")
doc = doc[["iso2", "Country_Region"]] # 특정 조건 = 중첩대괄호 !!! [[]]
doc # iso2는 왜 찾아왔냐 iso2 : 국가를 대표하는 시그니처 아이디 (국기명 찾아올 때 파일명으로 사용할 거라 필요함!)
```

```python
doc.duplicated() # 해당 인덱스값이 중복된 값이 존재한다면 True! 위에 3555~3559는 US로 중복되었기 때문에 True라고 뜨는 것
```

```python
doc[doc.duplicated()] # True인(=중복된) 값들만 뜨는 것임!

# 목적 : 현재 여러번 중복해서 사용된 값들을 1개만 남기고 나머지는 제거 (국가코드와 국가명은 1번씩만 나오면 되기 때문에 제거하는 것 같다!)
# 어떤 값을 살리고, 어떤 기준으로 나머지의 값을 제거할 것인가 (아래 셀)
```

```python
doc = doc.drop_duplicates(subset="Country_Region", keep="last")
# Country_Region가 기준이 되어야 하니까 부분집합에 넣어줌! subset=부분집합
# keep= 이라는 옵션 안 주면 기본이 first! last를 적어도 됨 근데 여기선 수치를 구하는 게 아니라서 그렇게 중요하진 않음
# dropna 처럼 drop듀 라고 기억하면 좋을듯!

doc # 3353 rows 에서 180 rows로 확 줄었음!!! US의 중복값이 많아서! china도! US는 마지막 행의 인덱스였던 3559만 남았음!
```
