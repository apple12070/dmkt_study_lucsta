```python
import pandas as pd
import plotly.graph_objects as go
import numpy as np

df = pd.DataFrame(np.random.rand(10, 2), columns=["A", "B"]) # np.random.rand(행, 열) : 2차원 배열 (*아래 셀에 예시 있음)

fig = go.Figure()

fig.add_trace(
    go.Bar(
        x=df.index, y=df["A"], name="A", text=df["A"], textposition="auto", texttemplate="%{y:.2f}"
    )
)
# textposition= auto, inside, outside
# texttemplate="%{y:.2f}" : y축 데이터를 소수둘째자리까지 float 형태로 나타내라 -> 소수점 너무 길 때 사용! 얘는 난수랏허 넘 길어

fig.add_trace(
    go.Bar(
        x=df.index, y=df["B"], name="B", text=df["B"], textposition="auto", texttemplate="%{y:.2f}"
    )
)

fig.update_layout( # .update_layout : 꾸미기
    {
        "title" : {
            "text": "Graph with <b>go.Bar</b>", # go.Bar : graph object의 Bar요소를 사용했다~ 딕셔너리 안에 <b> 요 마크업 인식되는 거 신기 ;;
            "x": 0.5,
            "y": 0.8,
            "font": {
                "size": 24
            }
        },
        "xaxis": {
            "title": "random number",
            "showticklabels": True,
            "dtick": 1 # 행(가로줄)의 눈금 간격!
        },
        "yaxis": {
            "title": "A",
            "dtick": 0.2 # 열(세로줄)의 눈금 간격!
        },
        "width": 900, # 총 가로 길이
        "height": 540 # 총 세로 길이
    }
)

fig.show()
```
