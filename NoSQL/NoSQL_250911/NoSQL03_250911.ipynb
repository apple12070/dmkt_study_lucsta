```python
import requests
from bs4 import BeautifulSoup

for page_num in range(10) :
    if page_num == 0 :
        res = requests.get("https://davelee-fun.github.io/")
    else :
        res = requests.get(f"https://davelee-fun.github.io/page{str(page_num + 1)}")
        # url은 문자열이어야 함. 정수와 문자열은 바로 더할 수 없기 때문. 그래서 str 씌워줌
    soup = BeautifulSoup(res.content, "html.parser")
    data = soup.select("div.card-body")
    for item in data :
        category = item.select_one("h2.card-title").get_text().replace("관련 상품 추천", "").strip()
        product = item.select_one("h4.card-text").get_text().replace("상품명", "").strip()
        # 브라우저는 div,h2 등의 구문 하나하나를 다 객체로 보고 있음. (Document Object Model)
        print(category, product)
```

```python
import requests
from bs4 import BeautifulSoup
from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017")
db = client["test"]
collection = db.daveshop

for page_num in range(10) :
    if page_num == 0 :
        res = requests.get("https://davelee-fun.github.io/")
    else :
        res = requests.get(f"https://davelee-fun.github.io/page{str(page_num + 1)}")
        # url은 문자열이어야 함. 정수와 문자열은 바로 더할 수 없기 때문. 그래서 str 씌워줌
    soup = BeautifulSoup(res.content, "html.parser")
    data = soup.select("div.card-body")
    for item in data :
        category = item.select_one("h2.card-title").get_text().replace("관련 상품 추천", "").strip()
        product = item.select_one("h4.card-text").get_text().replace("상품명", "").strip()
        # 브라우저는 div,h2 등의 구문 하나하나를 다 객체로 보고 있음. (Document Object Model)
        print(category, product)
        data = dict()
        data["category"] = category
        data["product"] = product
        collection.insert_one(data)

client.close()
```

```python
##### 문제 3 #####
# "매트리스커버" 카테고리의 모든 상품을 조회해주세요.

from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017")
db = client["test"]
collection = db.daveshop

products = collection.find(
    {"category": "매트리스커버"},
    {"_id": 0, "product": 1}  
)

for product in products:
    print(product["product"])

client.close()
```

```python
### 문제 3 - 해설 2 ###

from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017")
db = client["test"]
collection = db.daveshop

pipeline = [
    {"$match" : {"category" : "매트리스커버"}},
    {"$project" : {"_id" : 0, "product" : 1}}
]

for product in collection.aggregate(pipeline) :
    print(product["product"])
    
client.close()
```

```python
##### 문제 4 #####
# 각 카테고리별 상품 수를 조회해주세요.

from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017")
db = client["test"]
collection = db.daveshop

pipeline = [
    {"$group" : {"_id" : "$category", "count" : {"$sum" : 1}}},
    {"$project" : {"_id" : 0, "category" : "$_id", "count" : 1}}
]

for category_count in collection.aggregate(pipeline) :
    print(category_count["category"], category_count["count"])

client.close()
```

```python
##### 문제 5 #####
# 상품명에 "핑크"가 포함된 상품을 조회해주세요.

from pymongo import MongoClient

client = MongoClient("mongodb://localhost:27017")
db = client["test"]
collection = db.daveshop

pink_products = collection.find(
    {"product": {"$regex": ".*핑크.*"}}, 
    {"_id": 0, "product": 1} 
)

for p in pink_products:
    print(p["product"])

client.close()
```
