#### 8. MongoDB Aggregation FrameWork

```
- MongoDB > NoSQL, Aggregation=> 집합 | 집계
- Aggregation => 프레임워크
- $push <-> 배열이 아닌 요소들을 하나의 배열의 자료구조 형태로 만들어주는 기능
  * 각 문서의 필드 값을 배열로 추가하므로, 그룹 안의 모든 값이 누적됨
- 프레임워크 | 라이브러리 => 프레임워크는 독자적인 문법 사용 가능
- 기존에 학습했던 find() 함수의 상위호환 버전이라고 생각해도 무방!!!
- Linux 운영체제 => pipeline 개념을 벤치마킹 => 함수의 기능이 구현 => 기능
- shard : 샤드 => A (*1번 샤드) 종료 -> B (*2번 샤드) 
- shard - 해당 파이프라인의 중간단계에 있는 독립적인 하나의 요소요소. 샤드들이 모여 하나의 파이프라인이 완성됨.
- Aggregation FrameWork를 실행하기 위한 주요 단계
  - $addFields : 기존 필드에 추가적으로 속성을 더하거나 변경하고자 할 때 
  - $match : 조건에 맞는 문서를 필터링
  - $unwind : 중첩배열의 값을 하나로 합쳐주는 역할 (*$group 위에 $unwind는 문법처럼 붙어다님!!) / 배열 안에 있는 값들을 가져올 수 있도록 해주는 역할
  - $group : 문서를 그룹화하고 집계 작업을 수행할 수 있도록 준비
  - $project : 연산처리된 결과를 특정 요소에 포함시키거나 새로운 문서로 변환 작업
    (*연산처리를 하기 위한 집계 작업 명령어 = 예약어)
  - $limit : 특정 개수를 지정해서 출력하고자 할 때
-----
  - $sort : 연산처리된 결과를 정렬 (*위치 무관)
  - $skip : 특정 연산처리 후 의도적으로 문서를 건너뛰고 값을 찾아오려고 할 때
  - $count : 연산결과값을 통해서 받아온 문서 수의 개수를 집계
  - $lookup : 서로 다른 컬렉션을 묶어서 편집할 때 사용 (*JOIN)
  - $documents : 임시메모리 공간에 문서를 생성해놓을 때 사용
  - $merge : 임시 메모리 공간에 저장된 문서를 특정 컬렉션 안에 삽입
    (*into : 삽입할 컬렉션 명)
  - $facet : 어떤 하나의 측면 => interface : aggregation 프레임워크를 사용할 때, 단점을 보완할 수 있음 => 파이프라인 => 앞에서 설정해놓은 단계가 종료되어야 그 다음 단계가 실행
        (*만약, match 연산을 통해서 찾아와야 하는 조건식이 100줄 // group 3줄 => 파이프라인 내 상호간 순서에 영향을 받지 않는 요소라고 한다면, 동시 실행을 시킬 수 있음, 앞쪽에서 실행되어야할 요소를 기다리지 않고 병렬로 같이 실행!!!) => 그룹을 생성, A그룹 & B그룹
  - $redact : 사전적 의미 => "수정하다" (*어떤 조건에 따라서 수정 여부를 판단)
  - $cond : 조건을 따져오는 역할
-----
-> 샤드 내부에서 자체적으로 집계 및 정의를 하는 경우 (연산처리 없음) 예약어를 먼저 걸고 필드명 정의를 뒤에서 함(위의 예약어들은 예약어가 먼저 나옴)
ex) db.movies.aggregate([
  {$match: {year: {$gte: 2000}}},
  {$count: "movies_since_2000"}  <<------------------
])

-------------------------------------------------------------------------------------------------
-> 샤드 내부에서 그룹핑이 된 이후 (연산처리 있음) 정의를 할 때는 필드명을 앞에서 정의 (아래 연산함수들은 필드명 먼저 정의)
ex) db.movies.aggregate([
  {
    $group : {
      _id: "$year",
      titles: {$push: "$title"}  <<-----------------
    }
  }
])

    - $sum : 특정 필드명 안에 포함되어있는 값의 합계
    - $avg : 특정 필드명 안에 포함되어있는 값의 평균
    - $min : 특정 필드명 안에 포함되어있는 최솟값
    - $max : 특정 필드명 안에 포함되어있는 최댓값
    - $first : 특정 필드 안에 첫번째 값
    - $last : 특정 필드 안에 마지막번째 값
    - $push : 기존 값들을 배열의 형태로 구성할 때 사용
    - $convert : 특정 필드 내 속성을 변환시킬 때 사용
    - $strLenCP : 문자열의 길이값을 찾아올 때 사용
    - $toString : 문자열 형태가 아닌 타입의 값을 문자로 변환
    - $addToSet : 특정 필드 안에 있는 값을 중복하지 않고 찾아오고자 할 때
    - $concat : 각 필드 안에 존재하는 값을 하나의 문자열로 생성하고자 할 때


* 필드명 작성 순서
1. $match, $group : 해당 명령어 안에 어떤 조건식을 통해서 결과값이 도출되었음. 그리고 해당 결과값을 어떤 필드에 값을 담아야 하는 경우
  => 무조건 내부적으로 필드명을 먼저 사용!!!
2. $count, $sum, $avg : 어떤 값들을 집계한 후 새로운 필드를 생성할 때, 이때에는 무조건 예약어(*명령어)를 먼저 사용 후 이후에 속성값으로 필드명을 입력

```
